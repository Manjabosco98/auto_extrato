```python
import os
import shutil
import pandas as pd
from pathlib import Path

CAMINHOS = r"G:\.shortcut-targets-by-id\1w3AY5xp-hHyQh263jfvhGoUdzH-0Ls0F\AUTOEXT\DOCS E CAMINHO.xlsx"
EMP_ATIVOS = r"C:\Users\manja\OneDrive\Documentos\AutoExtrato\data\BaseEmpAtivas.xlsm"
DOCS = Path(r"G:\.shortcut-targets-by-id\1w3AY5xp-hHyQh263jfvhGoUdzH-0Ls0F\AUTOEXT\DOCS")
EMP = (r"G:\.shortcut-targets-by-id\1w3AY5xp-hHyQh263jfvhGoUdzH-0Ls0F\AUTOEXT\DOCS\EMP")

caminhos = pd.read_excel(CAMINHOS)

if DOCS.exists():
    arquivos = [item for item in DOCS.iterdir() if item.is_file() and item.name.lower() != "desktop.ini"]
    for arquivo in arquivos:
        mes = arquivo.stem.split("_")[0][:2]
        ano = arquivo.stem.split("_")[0][-2:]
        codigo = arquivo.stem.split("_")[1]
        emp = arquivo.stem.split("_")[-1].upper()

        #  Buscando empresa na base de ativos
        emp_ativos = pd.read_excel(EMP_ATIVOS, sheet_name="EmpAtivas")
        emp_ativos["chave"] = emp_ativos.apply(lambda x: f"{x['ID']}_{x['Razão social']}", axis=1)
        emp_ativos = emp_ativos[["ID", "Razão social", "chave", "CNPJ", "Status"]]
        id_emp = emp_ativos.loc[emp_ativos["Razão social"] == emp, "chave"].iloc[0]
        
        # Buscando caminho na base de DOCS
        destino = caminhos.loc[caminhos["Codigo"] == str(codigo), "destino_final"].iloc[0]
        destino = Path((destino.replace("{EMPRESA}", str(id_emp)).replace("{ANO}", str(ano)).replace("{MES}", str(mes))))
        destino_final = DOCS / destino / arquivo.name
        destino_final.parent.mkdir(parents=True, exist_ok=True)

        shutil.move(str(arquivo), str(destino_final))
        display(destino_final)
```
